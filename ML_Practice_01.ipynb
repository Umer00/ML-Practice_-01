# 📘 Introduction to Machine Learning

### 🎯 Objective
In this lecture, we will understand the basics of Machine Learning, explore its real-world applications, and build our first ML model (Linear Regression).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

### Step 1: Create Dataset

```python
data = {
    'Experience' : [4,3,2,3,2,1.5,1,4.5,4,2,2,1,3.5,3,1],
    'Salary' : [500000,300000,200000,300000,200000,150000,100000,550000,500000,200000,200000,100000,350000,300000,100000]
}
df=pd.DataFrame(data)
df
```

```python
df.shape
```

### Step 2: Visualize the Data

```python
plt.scatter(df['Experience'],df['Salary'],color='blue')
plt.xlabel("How much experince person have")
plt.ylabel("salary")
plt.show()
```

### Step 3: Split Data into Training and Testing

```python
x = df[['Experience']]
y = df[['Salary']]

x_train, x_test, y_train, y_test =train_test_split(x,y, test_size=0.3,random_state=32)
```

```python
x_train
```

```python
x_test
```

```python
x_train.shape , x_test.shape
```

```python
y_test
```

### Step 4: Train Linear Regression Model

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_predict= model.predict(x_test)
y_predict
```

```python
y_test
```

```python
print("Mean Scored Error: ", mean_squared_error(y_test,y_predict))
print("R2_score: ", r2_score(y_test,y_predict))
```

```python
plt.scatter(x,y,color='blue')
plt.plot(x,model.predict(x),color='red')
plt.xlabel("How much experince person have")
plt.ylabel("salary")
plt.title("Linear Regrssion Fit")
plt.show()
```
